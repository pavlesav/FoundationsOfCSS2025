Now we want to plot a different way using matplotlib  
https://matplotlib.org/stable/index.html

```python
x = range(10)
y = range(10)
y = [0]*10
y = [0]*5+[2]*5
plt.plot(x,y,label='random data')
plt.title('this is a title')
plt.legend()
```

now we want to use a different color, linestyle, and marker  
https://www.geeksforgeeks.org/python/line-plot-styles-in-matplotlib/

```python
plt.plot(x,y,label='random data',color='b',linestyle='--', marker='v')
plt.legend()
```

Now we want to create a scatter plot.

```python
x = np.random.randint(low=0,high=25,size=50)
y = x*2+np.random.randint(-5,5,50)
plt.scatter(x=x,y=y,label='random',color='r',marker='v')
plt.legend()
```

Now we want to add horizontal and vertical lines to our data.

```python
x = np.random.randint(low=0,high=25,size=50)
y = x*2+np.random.randint(-5,5,50)
plt.scatter(x=x,y=y,label='random',color='r',marker='v')
plt.axhline(20)
plt.axvline(10)
plt.hlines(21,5,20)
plt.legend()
```

Now we want to add a specific line to the plot.

```python
x = np.random.randint(low=0,high=25,size=50)
y = x*2+np.random.randint(-5,5,50)
plt.scatter(x=x,y=y,label='random',color='r',marker='v')
plt.plot([5,20],[10,40])
plt.legend()
```

Now we want to add text to our plot.

```python
x = np.random.randint(low=0,high=25,size=50)
y = x*2+np.random.randint(-5,5,50)
plt.scatter(x=x,y=y,label='random',color='r',marker='v')
plt.text(x=5,y=20,s='test',size=100)
plt.legend()
```

Now we want to create a bar plot.

```python
x = [1,2,3]
y = [10,15,20]
plt.bar(x=x,height=y)
```

Now we want to create multiple plots in one plot.

```python
figure, axis = plt.subplots(2,2)
axis[0,0].plot(x,y)
axis[0,1].bar(x,y)
axis[1,0].scatter(x,y)
axis[1,1].plot(y,np.array(x)**5)
figure.suptitle('alot of plots')
```

Now we want to save our plot as png.

```python
figure, axis = plt.subplots(2,2)
axis[0,0].plot(x,y)
axis[0,1].bar(x,y)
axis[1,0].scatter(x,y)
axis[1,1].plot(y,np.array(x)**5)
figure.suptitle('alot of plots')
figure.savefig('cool_plot.png',transparent=True)
```
